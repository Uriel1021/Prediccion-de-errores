***Importar livrerias***

```python
import pandas as pd
import numpy as np
from scipy.io import arff
import matplotlib.pyplot as plt
import seaborn as sns
```

***Importar datos***

```python
data, meta = arff.loadarff("../data/raw/kc1.arff")

df = pd.DataFrame(data)
```

```python
df.head(2)
```

***Preparar los datos***

```python
numericas = df.select_dtypes(include=np.number).columns.tolist()
```

```python
df[numericas].describe().T
```

```python
correlacion = df[numericas].corr()
```

```python
plt.Figure()
sns.heatmap(correlacion)
plt.title("Matriz de correlacion de los datos numericos de KC1")
```
